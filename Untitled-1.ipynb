```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler

# ====================================================================
# WEEK 1: DATA ACQUISITION, PREPROCESSING & EXPLORATORY DATA ANALYSIS
# Intern Name: Sharmila V
# Domain: Data Science Virtual Internship (Yuva Intern)
# ====================================================================

# Step 1: Ingest Data
print("--> Ingesting raw Titanic dataset...")
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

print(f"Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns")
print("\nMissing values before cleaning:")
print(df.isnull().sum()[df.isnull().sum() > 0])

# Step 2: Feature Engineering & Imputation
# Regular expression title parsing
df['Title'] = df['Name'].str.extract(' ([A-Za-z]+)\.', expand=False)
df['Title'] = df['Title'].replace(['Lady', 'Countess','Capt', 'Col', 'Don', 'Dr', 'Major', 'Rev', 'Sir', 'Jonkheer', 'Dona'], 'Rare')
df['Title'] = df['Title'].replace(['Mlle', 'Ms'], 'Miss').replace('Mme', 'Mrs')

# Conditional median imputation for Age based on Pclass and Title
df['Age'] = df.groupby(['Pclass', 'Title'])['Age'].transform(lambda x: x.fillna(x.median()))
df['Age'] = df['Age'].fillna(df['Age'].median())

# Mode imputation for Embarked
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Binary indicator for Cabin privilege
df['Has_Cabin'] = df['Cabin'].apply(lambda x: 0 if pd.isna(x) else 1)
df.drop(columns=['Cabin'], inplace=True)

# Relational family features
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
df['IsAlone'] = (df['FamilySize'] == 1).astype(int)

# Outlier handling for Fare via IQR capping and Log transformation
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)
IQR = Q3 - Q1
upper_whisker = Q3 + 1.5 * IQR
fare_99th = df['Fare'].quantile(0.99)
df['Fare_Capped'] = np.where(df['Fare'] > fare_99th, fare_99th, df['Fare'])
df['Fare_Log'] = np.log1p(df['Fare'])

# Step 3: Design Matrix & Scaling
feature_cols = ['Pclass', 'Sex', 'Age', 'Fare_Log', 'FamilySize', 'IsAlone', 'Embarked', 'Title', 'Has_Cabin']
X = pd.get_dummies(df[feature_cols], drop_first=True, dtype=int)
scaler = StandardScaler()
X[['Age_Scaled', 'Fare_Scaled']] = scaler.fit_transform(X[['Age', 'Fare_Log']])

print("\nMissing values after cleaning:")
print(f"Total remaining nulls: {df.isnull().sum().sum()}")
print(f"Preprocessed design matrix shape: {X.shape}")

# Step 4: Visualizations
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Subplot 1: Gender survival rate
sns.barplot(data=df, x='Sex', y='Survived', palette='Blues_d', ax=axes[0, 0])
axes[0, 0].set_title('Survival Rate by Gender (74.2% Female vs 18.9% Male)', fontweight='bold')
axes[0, 0].set_ylabel('Survival Probability')

# Subplot 2: Socio-economic Class survival
sns.barplot(data=df, x='Pclass', y='Survived', palette='Greens_d', ax=axes[0, 1])
axes[0, 1].set_title('Survival Rate by Passenger Class (63.0% Class 1 vs 24.2% Class 3)', fontweight='bold')
axes[0, 1].set_ylabel('Survival Probability')

# Subplot 3: Fare Distribution Before Log Transformation
sns.histplot(df['Fare'], kde=True, color='red', ax=axes[1, 0], bins=30)
axes[1, 0].set_title('Raw Fare Skewness (Kurtosis > 30)', fontweight='bold')
axes[1, 0].set_xlabel('Raw Ticket Fare ($)')

# Subplot 4: Normalized Log Fare
sns.histplot(df['Fare_Log'], kde=True, color='green', ax=axes[1, 1], bins=30)
axes[1, 1].set_title('Normalized Log-Transformed Fare (Gaussian Bell Curve)', fontweight='bold')
axes[1, 1].set_xlabel('Log Transformed Fare: ln(1 + Fare)')

plt.tight_layout()
plt.savefig('preprocessing_eda_plots.png', dpi=300)
plt.show()
print("--> Process Complete! Saved plot: preprocessing_eda_plots.png")
```
